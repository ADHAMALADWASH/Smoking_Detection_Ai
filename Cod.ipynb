```python
from ultralytics import YOLO
```

```python
model = YOLO('yolov8n.pt')
```

```python
import yaml

data = {
    'train': './images',
    'val': './images',
    'nc': 1,
    'names': ['cigarette']
}

with open('data.yaml', 'w') as f:
    yaml.dump(data, f, default_flow_style=False, sort_keys=False)

print("Yaml Done ")
```

```python
results = model.train(
    data='data.yaml', 
    epochs=50,
    imgsz=640,
    batch=16,
    device=0 ,
    optimizer='AdamW',   
)
print("Model Results")
print(f"Accuarcy (mAP@50): {results.box.map50 * 100:.2f}%")
```

```python
import cv2
import winsound  # مكتبة إصدار الأصوات المدمجة في ويندوز
from ultralytics import YOLO

# 1. تحميل الموديل الخاص بك
model_path = r'C:\Users\AdminOS\runs\detect\train-3\weights\best.pt'
model = YOLO(model_path)

# 2. فتح الكاميرا
cap = cv2.VideoCapture(0)

print("جاري تشغيل الكاميرا... اضغط على حرف 'q' للإغلاق.")

while cap.isOpened():
    success, frame = cap.read()
    if not success:
        print("تعذر الوصول للكاميرا.")
        break

    # 3. إرسال الصورة للموديل
    results = model(frame, conf=0.4)

    # 4. التحقق من وجود السجائر
    if len(results[0].boxes) == 0:
        # الحالة الأولى: لم يكتشف أي سيجارة (أمان)
        cv2.putText(frame, "STATUS: SAFE", (30, 50), 
                    cv2.FONT_HERSHEY_SIMPLEX, 1.2, (0, 255, 0), 3)
        annotated_frame = frame
        
    else:
        # الحالة الثانية: اكتشف سيجارة (إنذار)
        annotated_frame = results[0].plot()
        cv2.putText(annotated_frame, "STATUS: SMOKING DETECTED!", (30, 50), 
                    cv2.FONT_HERSHEY_SIMPLEX, 1.2, (0, 0, 255), 3)
        
        # إطلاق صوت إنذار (الرقم الأول هو التردد بالهرتز، والثاني هو المدة بالملي ثانية)
        winsound.Beep(2500, 150) 

    # 5. عرض النافذة
    cv2.imshow("Smoking Detection System", annotated_frame)

    # 6. إغلاق البرنامج عند الضغط على حرف 'q'
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
